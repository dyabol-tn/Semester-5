Задание 1. На отрезке [0,2] получить приближенное решение задачи Коши,
используя модуль scipy.integrate. Вывести график решения.


```python
import numpy as  np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp, solve_bvp

def dydt(t, y):
    dy0dt = y[1]
    dy1dt = y[2]
    dy2dt = y[2] * 2 * np.exp(-t) - y[1] - 2 * (y[0] ** 2)
    return dy0dt, dy1dt, dy2dt

t = np.linspace(0, 2, 101)
sol = solve_ivp(dydt, (0, 2), [0, 0, 1], t_eval = t)
plt.plot(sol.t, sol.y[0])
plt.show()
```

Задание 2. Получить приближенное решение краевой задачи, используя модуль
scipy.integrate. Вывести график решения.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp, solve_bvp

def dudx(x, u):
    du0dx = u[1]
    du1dx = (1.0 / x) * u[1] + x * u[0]
    return du0dx, du1dx

def bc_fun(ua, ub):
    a = ua[1] - 2.0
    b = ub[0] + ub[1] - 1.0
    return a, b

x = np.linspace(0.5, 1, 101)
u = np.zeros((2, x.size))
sol = solve_bvp(dudx, bc_fun, x, u)
plt.plot(sol.x, sol.y[0])
plt.show()
```
