Вариант 21. Дано натуральное число. Определить, является ли оно равноудаленным от двух ближайших простых чисел. Выведите true или false.

```python
import math

a = 10007
search_left = a - 1
search_right = a + 1
dist_left = 1
dist_right = 1
check_left = False
check_right = False

if search_left < 2:
    dist_left = 0
else:
    i = 2
    while i <= math.isqrt(search_left):
        if search_left % i == 0:
            dist_left += 1
            check_left = True
            search_left -= 1
            i = 2
        else:
            i += 1
i = 2
while i <= math.isqrt(search_right):
    if search_right % i == 0:
        dist_right += 1
        check_right = True
        search_right += 1
        i = 2
    else:
        i += 1
if dist_left == dist_right and dist_left > 0:
    print("true")
else:
    print("false")
```

Вариант 21. В списке перепишите все ненулевые элементы в начало списка (сохраняя порядок), а нулевые - в конец.

```python
lst =  [2, 0 , 5, 7, 8, 0, 0, 0, 15, 0, 65]
position = 0
print(lst)
for i in range (len(lst)):
    if lst[i] != 0:
        lst[position] = lst[i]
        position += 1
for i in range (position, len(lst)):
    lst[i] = 0
print(lst)
```
